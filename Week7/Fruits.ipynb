```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
# 1. Load the Fruit dataset directly from public repository source
url = "https://raw.githubusercontent.com/saanyalasod/Fruits/main/fruit%20data.csv"
```

```python
# Read the data into a DataFrame
df = pd.read_csv(url)
```

```python
# Clean up data: handle any trailing empty rows or missing values safely
df = df.dropna()
```

```python
# Map features (X) and target label (y)
# The dataset has columns: fruit_label, fruit_name, fruit_subtype, mass, width, height, color_score
X = df[['mass', 'width', 'height', 'color_score']].values
y = df['fruit_label'].values
```

```python
# Track fruit name mapping to show class references clearly
fruit_mapping = dict(zip(df['fruit_label'].astype(int), df['fruit_name']))
```

```python
# Define the experimental parameters
test_sizes = {
    "90-10 Split": 0.10,
    "70-30 Split": 0.30
}
k_values = [3, 5, 7]

print(f"Dataset Loaded Successfully! Total fruit samples: {X.shape[0]}")
print(f"Fruit classes detected: {fruit_mapping}")
print("--- KNN Fruit Classification Results ---")

# 2. Loop through each train-test split configuration
for split_name, test_size_val in test_sizes.items():
    print(f"\n=========================================")
    print(f" Evaluation for {split_name}")
    print(f"=========================================")
    
    # Split the dataset
    # 'stratify=y' ensures each split has a fair balance of apples, lemons, oranges, etc.
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size_val, random_state=42, stratify=y
    )
    
    print(f"Training samples: {X_train.shape[0]} | Testing samples: {X_test.shape[0]}")
    print("-" * 50)
    
    # Feature Scaling: Highly critical since 'mass' (e.g., 150g) is numerically much larger 
    # than 'width' (e.g., 7.4 cm) or 'color_score' (e.g., 0.7)
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # 3. Loop through each K value for the current split
    for k in k_values:
        # Initialize the KNN classifier
        knn = KNeighborsClassifier(n_neighbors=k)
        
        # Train the model using the scaled dataset
        knn.fit(X_train_scaled, y_train)
        
        # Make predictions on the scaled test dataset
        y_pred = knn.predict(X_test_scaled)
        
        # Calculate accuracy score
        accuracy = accuracy_score(y_test, y_pred)
        
        print(f"K = {k} -> Accuracy: {accuracy:.4f}")
```
