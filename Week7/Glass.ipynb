```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
# 1. Load the Glass Identification Dataset directly from UCI Repo URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/glass/glass.data"
```

```python
# Define the columns according to the UCI dataset documentation
column_names = ["Id", "RI", "Na", "Mg", "Al", "Si", "K", "Ca", "Ba", "Fe", "Type_of_Glass"]
```

```python
# Read the data into a DataFrame
df = pd.read_csv(url, names=column_names)
```

```python
# Drop the 'Id' column as it's just an index identifier
df = df.drop(columns=["Id"])
```

```python
# Separate features (X) and target label (y)
X = df.drop(columns=["Type_of_Glass"]).values
y = df["Type_of_Glass"].values
```

```python
# Define the experimental parameters
test_sizes = {
    "90-10 Split": 0.10,
    "70-30 Split": 0.30
}
k_values = [3, 5, 7]

print(f"Dataset Loaded Successfully! Total instances: {X.shape[0]}, Total features: {X.shape[1]}")
print("--- KNN Glass Classification Results ---")

# 2. Loop through each train-test split configuration
for split_name, test_size_val in test_sizes.items():
    print(f"\n=========================================")
    print(f" Evaluation for {split_name}")
    print(f"=========================================")
    
    # Split the dataset
    # Note: 'stratify=y' ensures your train/test splits have proportional representation of glass types
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size_val, random_state=42, stratify=y
    )
    
    print(f"Training samples: {X_train.shape[0]} | Testing samples: {X_test.shape[0]}")
    print("-" * 50)
    
    # Feature Scaling: Crucial for distance calculation in KNN
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # 3. Loop through each K value for the current split
    for k in k_values:
        # Initialize the KNN classifier
        knn = KNeighborsClassifier(n_neighbors=k)
        
        # Train the model using the SCALED data
        knn.fit(X_train_scaled, y_train)
        
        # Make predictions on the scaled test set
        y_pred = knn.predict(X_test_scaled)
        
        # Calculate accuracy
        accuracy = accuracy_score(y_test, y_pred)
        
        print(f"K = {k} -> Accuracy: {accuracy:.4f}")
```
