```python
# HILL CLIMBING ALGORITHM

# Hill Climbing is an optimization algorithm used in Artificial Intelligence.
# It starts from one solution and keeps moving to a better neighboring solution.

# Formula used in this program:
# f(x) = -(x^2) + 10x

# Goal:
# Find the value of x where the function gives maximum value.

# Example:
# If x = 5
# f(5) = -(5^2) + 10(5)
#      = -25 + 50
#      = 25

# The algorithm checks neighboring values like:
# x-1 and x+1

# It moves toward the better value until no improvement is possible.
```

```python
# Function to calculate f(x)

def f(x):
    return -(x**2) + 10*x
```

```python
# User enters starting value

current = int(input("Enter starting value of x: "))
```

```python
# Hill Climbing Process

iteration = 1

while True:

    # Calculate left and right neighbors
    left = current - 1
    right = current + 1

    # Calculate function values
    current_value = f(current)
    left_value = f(left)
    right_value = f(right)

    # Print iteration details
    print("\nIteration :", iteration)

    print("Current x =", current)
    print("f(x) =", current_value)

    print("Left Neighbor =", left, " Value =", left_value)
    print("Right Neighbor =", right, " Value =", right_value)

    # Move to better neighbor
    if left_value > current_value:
        current = left

    elif right_value > current_value:
        current = right

    else:
        # Stop if no better neighbor exists
        print("\nBest solution found!")
        print("Best x =", current)
        print("Maximum value =", current_value)
        break

    iteration += 1
```
