```python
# Import numerical computing library
import numpy as np

# Import data handling library
import pandas as pd

# Import plotting library
import matplotlib.pyplot as plt

# Import advanced statistical visualization library
import seaborn as sns

# Import sample dataset
from sklearn.datasets import load_iris

# Import 3D plotting toolkit
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Load iris dataset from sklearn
iris = load_iris()

# Convert dataset into DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add target column to dataframe
df['species'] = iris.target

# Display first 5 rows
print(df.head())
```

```python
# Set figure size
plt.figure(figsize=(8,6))

# Create scatter plot
plt.scatter(
    df['sepal length (cm)'],   # X-axis values
    df['sepal width (cm)'],    # Y-axis values
    c=df['species'],           # Color based on species
    cmap='viridis'             # Color style
)

# Add title
plt.title("Scatter Plot of Sepal Length vs Sepal Width")

# Label X-axis
plt.xlabel("Sepal Length")

# Label Y-axis
plt.ylabel("Sepal Width")

# Show plot
plt.show()
```

```python
# Set figure size
plt.figure(figsize=(8,6))

# Create box plot
sns.boxplot(data=df.iloc[:,0:4])

# Add title
plt.title("Box Plot of Iris Features")

# Show plot
plt.show()
```

```python
# Calculate correlation matrix
corr = df.iloc[:,0:4].corr()

# Set figure size
plt.figure(figsize=(8,6))

# Create heatmap
sns.heatmap(
    corr,              # Correlation data
    annot=True,        # Show values inside boxes
    cmap='coolwarm'    # Color style
)

# Add title
plt.title("Correlation Heat Map")

# Show plot
plt.show()
```

```python
# Create x values
x = np.linspace(-5, 5, 100)

# Create y values
y = np.linspace(-5, 5, 100)

# Create mesh grid
X, Y = np.meshgrid(x, y)

# Define mathematical function
Z = np.sin(np.sqrt(X**2 + Y**2))
```

```python
# Set figure size
plt.figure(figsize=(8,6))

# Create contour plot
contour = plt.contourf(
    X, Y, Z,      # Grid data
    cmap='plasma' # Color map
)

# Add color bar
plt.colorbar(contour)

# Add title
plt.title("Contour Plot")

# Show plot
plt.show()
```

```python
# Create figure
fig = plt.figure(figsize=(10,7))

# Add 3D axis
ax = fig.add_subplot(111, projection='3d')

# Create surface plot
surface = ax.plot_surface(
    X, Y, Z,            # Surface coordinates
    cmap='viridis'      # Color style
)

# Add title
ax.set_title("3D Surface Plot")

# Label X-axis
ax.set_xlabel("X Axis")

# Label Y-axis
ax.set_ylabel("Y Axis")

# Label Z-axis
ax.set_zlabel("Z Axis")

# Add color bar
fig.colorbar(surface)

# Show plot
plt.show()
```
