```python
# GREEDY BEST-FIRST SEARCH (GBFS)

# GBFS is an AI search algorithm that uses a heuristic function h(n) 
# to estimate the distance from the current node to the goal.

# Key Logic:
# It always picks the neighbor that "looks" closest to the goal.
# f(n) = h(n)

# Goal:
# Reach the 'Goal' node using the most promising path based on heuristics.
```

```python
# Function to take dynamic user input for the search space
def get_user_data():
    heuristics = {}
    graph = {}

    print("--- STEP 1: DEFINE HEURISTICS ---")
    nodes_count = int(input("How many nodes are in your graph? "))
    for _ in range(nodes_count):
        node = input("Enter node name: ").strip()
        h_val = int(input(f"Enter heuristic value for {node} (dist to goal): "))
        heuristics[node] = h_val
        graph[node] = []

    print("\n--- STEP 2: DEFINE CONNECTIONS ---")
    edge_count = int(input("How many connections (edges) exist? "))
    for _ in range(edge_count):
        u, v = input("Enter connection (format: Node1 Node2): ").split()
        graph[u].append(v)
        # Assuming undirected? If directed, comment out the next line:
        # graph[v].append(u) 

    return graph, heuristics

# Initialize the data
graph, heuristics = get_user_data()
```

```python
# User enters the mission parameters
start_node = input("\nEnter the starting node: ").strip()
goal_node = input("Enter the goal node: ").strip()

if start_node not in graph or goal_node not in graph:
    print("Error: One of the nodes entered does not exist in your graph definition.")
```

```python
import heapq

# The core search logic
def gbfs(start, goal, graph, heuristics):
    # Priority Queue: (heuristic, current_node, path)
    pq = [(heuristics[start], start, [start])]
    visited = set()
    iteration = 1

    while pq:
        h, current, path = heapq.heappop(pq)

        if current in visited:
            continue

        print(f"\nIteration {iteration}: Visiting {current}")
        print(f"Heuristic of {current} is {h}")

        if current == goal:
            print("\n--- GOAL REACHED! ---")
            return path

        visited.add(current)

        for neighbor in graph.get(current, []):
            if neighbor not in visited:
                heapq.heappush(pq, (heuristics[neighbor], neighbor, path + [neighbor]))
        
        iteration += 1
    
    return None

# Execute search
final_path = gbfs(start_node, goal_node, graph, heuristics)

if final_path:
    print("Final Path found:", " -> ".join(final_path))
else:
    print("No path exists between these nodes.")
```
